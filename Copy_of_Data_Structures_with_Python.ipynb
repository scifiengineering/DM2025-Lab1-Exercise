## List Manager

```python
class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def add_node(self, val):
        """Adds a new node to the end of the list (Append)."""
        new_node = ListNode(val)
        if not self.head:
            self.head = new_node
            return

        # Traverse to the end of the list
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def delete_node_by_key(self, key):
        """Removes the first node containing the specified value."""
        current = self.head

        # 1. If the list is empty
        if not current:
            return

        # 2. If the head node itself holds the key
        if current.val == key:
            self.head = current.next
            return

        # 3. Search for the key while keeping track of the previous node
        prev = None
        while current and current.val != key:
            prev = current
            current = current.next

        # 4. If key was not found
        if not current:
            print(f"Key '{key}' not found.")
            return

        # 5. Unlink the node
        prev.next = current.next

    def delete_node_by_position(self, position):
        """Removes the node at the specific index (0-based)."""
        if not self.head:
            return

        current = self.head

        # 1. If deleting the head (position 0)
        if position == 0:
            self.head = current.next
            return

        # 2. Find the node BEFORE the one to be deleted
        prev = None
        count = 0
        while current and count < position:
            prev = current
            current = current.next
            count += 1

        # 3. If position is out of bounds
        if not current:
            print("Position out of range.")
            return

        # 4. Unlink the target node
        prev.next = current.next

    def get_list_as_string(self):
        """Helper for debugging: returns the full path of the list."""
        nodes = []
        current = self.head
        while current:
            nodes.append(str(current.val))
            current = current.next
        return " -> ".join(nodes)
```

```python
my_list = SinglyLinkedList()
for i in range(1, 11):
  my_list.add_node(i)
my_list.get_list_as_string()
```

# List Viewer

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

class TimelineGUI:
    def __init__(self, list_manager, item_intro="📍 Current Node:"):
        self.manager = list_manager
        self.current = list_manager.head
        self.item_intro = item_intro

        self.output = widgets.Output()
        self.next_button = widgets.Button(description="Next ➡️")
        self.prev_button = widgets.Button(description="⬅️ Prev")

        self.next_button.on_click(self.on_next_clicked)
        self.prev_button.on_click(self.on_prev_clicked)

    def render(self):
        controls = widgets.HBox([self.prev_button, self.next_button])
        display(controls, self.output)
        self.refresh_text()

    def refresh_text(self):
        with self.output:
            clear_output(wait=True)
            if not self.current:
                print("📭 The list is empty.")
                return

            print(f"{self.item_intro}")

            # --- Flexible Data Display Logic ---
            data = self.current.val

            if isinstance(data, dict):
                # If it's a dictionary (inventory), loop through key-value pairs
                for key, value in data.items():
                    print(f"  • {key.capitalize()}: {value}")
            else:
                # If it's a simple value (int, str), just print it
                print(f"  {data}")

            # --- Navigation Logic ---
            is_circular = getattr(self.manager, 'is_circular', False)
            if not is_circular:
                self.next_button.disabled = self.current.next is None
                self.prev_button.disabled = not hasattr(self.current, 'prev') or self.current.prev is None
            else:
                self.next_button.disabled = False
                self.prev_button.disabled = False

    def on_next_clicked(self, b):
        if self.current and self.current.next:
            self.current = self.current.next
            self.refresh_text()

    def on_prev_clicked(self, b):
        if self.current and hasattr(self.current, 'prev') and self.current.prev:
            self.current = self.current.prev
            self.refresh_text()
```

```python
class DoublyNode:
    def __init__(self, val):
        self.val = val
        self.next = None
        self.prev = None  # New pointer for backward navigation

class DoublyLinkedList:
    def __init__(self):
        self.head = None

    def add_node(self, val):
        """Adds a node to the end of the list. O(n)"""
        new_node = DoublyNode(val)
        if not self.head:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        # Two-way link
        current.next = new_node
        new_node.prev = current

    def delete_node_by_key(self, key):
        """Removes node by matching value."""
        current = self.head
        while current:
            if current.val == key:
                # If it's not the first node, update the previous node's next
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next # Deleting the head

                # If it's not the last node, update the next node's prev
                if current.next:
                    current.next.prev = current.prev
                return
            current = current.next

    def delete_node_by_position(self, position):
        """Removes node by index."""
        if not self.head:
            return

        current = self.head
        count = 0
        while current and count < position:
            current = current.next
            count += 1

        if current:
            if current.prev:
                current.prev.next = current.next
            else:
                self.head = current.next

            if current.next:
                current.next.prev = current.prev
```

```python
taiwan_timeline_doubly = DoublyLinkedList()
for year, era, headline, description in FULL_TAIWAN_HISTORY_TIMELINE_EXPANDED:
    # entry is (year, era, headline, description)
    historic_event = {
        "year": year,
        "era": era,
        "headline": headline,
        "description": description
    }

    taiwan_timeline_doubly.add_node(historic_event)
```
